# Parquet File Upload and Metadata Display

This notebook demonstrates how to upload a parquet file and display its metadata using firefly_client 3.4.0.

Note: This example creates a sample parquet file with astronomical data for demonstration purposes.


## Setup


Import required libraries:


```python
import pandas as pd
import numpy as np
from firefly_client import FireflyClient
import tempfile
import os
```

## Create Sample Parquet File


Create a sample parquet file with astronomical data for demonstration:


```python
# Create sample astronomical data
np.random.seed(42)
n_objects = 100

# Generate sample data
data = {
    'ra': np.random.uniform(0, 360, n_objects),  # Right Ascension in degrees
    'dec': np.random.uniform(-90, 90, n_objects),  # Declination in degrees
    'mag_g': np.random.normal(20, 2, n_objects),  # G-band magnitude
    'mag_r': np.random.normal(19.5, 1.8, n_objects),  # R-band magnitude
    'mag_i': np.random.normal(19, 1.5, n_objects),  # I-band magnitude
    'redshift': np.random.exponential(0.1, n_objects),  # Redshift
    'galaxy_type': np.random.choice(['spiral', 'elliptical', 'irregular'], n_objects),
    'distance': np.random.uniform(10, 1000, n_objects)  # Distance in Mpc
}

# Create DataFrame
df = pd.DataFrame(data)

# Add some metadata
df.attrs = {
    'title': 'Sample Galaxy Catalog',
    'description': 'Simulated galaxy catalog for firefly_client demonstration',
    'author': 'firefly_client demo',
    'creation_date': '2024-01-01',
    'coordinate_system': 'ICRS',
    'wavelength_units': 'Angstrom',
    'distance_units': 'Mpc'
}

print(f"Created DataFrame with {len(df)} objects")
print(f"Columns: {list(df.columns)}")
print(f"Metadata: {df.attrs}")
```

```python
# Save to parquet file
parquet_file = 'sample_galaxy_catalog.parquet'
df.to_parquet(parquet_file, index=False)
print(f"Saved parquet file: {parquet_file}")
print(f"File size: {os.path.getsize(parquet_file)} bytes")
```

## Initialize Firefly Client


Connect to a Firefly server. You can use the IRSA Viewer (public server) or a local Firefly instance:


```python
# Option 1: Use IRSA Viewer (public server)
url = 'https://irsa.ipac.caltech.edu/irsaviewer'

# Option 2: Use local Firefly server (uncomment if you have one running)
# url = 'http://127.0.0.1:8080/firefly'

# Create Firefly client
fc = FireflyClient.make_client(url)
print(f"Connected to Firefly server: {url}")
```

## Upload and Display Parquet File


Upload the parquet file to the Firefly server:


```python
# Upload the parquet file
file_on_server = fc.upload_file(parquet_file)
print(f"File uploaded successfully. Server location: {file_on_server}")
```

Display the table with metadata information:


```python
# Display the table with metadata
fc.show_table(
    file_on_server=file_on_server,
    tbl_id='galaxy_catalog',
    title='Sample Galaxy Catalog',
    page_size=20,
    is_catalog=True,  # This will overlay the objects on any images
    visible=True
)
print("Table displayed in Firefly viewer")
```

## Display Metadata Information


Show the metadata that was attached to the parquet file:


```python
# Display metadata information
print("=== PARQUET FILE METADATA ===")
for key, value in df.attrs.items():
    print(f"{key}: {value}")

print("\n=== DATA SUMMARY ===")
print(f"Number of objects: {len(df)}")
print(f"RA range: {df['ra'].min():.2f} - {df['ra'].max():.2f} degrees")
print(f"Dec range: {df['dec'].min():.2f} - {df['dec'].max():.2f} degrees")
print(f"G-band magnitude range: {df['mag_g'].min():.2f} - {df['mag_g'].max():.2f}")
print(f"Redshift range: {df['redshift'].min():.3f} - {df['redshift'].max():.3f}")

print("\n=== GALAXY TYPE DISTRIBUTION ===")
print(df['galaxy_type'].value_counts())
```

## Create Visualizations


Create some basic visualizations using the uploaded data:


```python
# Create a color-magnitude diagram
trace1 = {
    'tbl_id': 'galaxy_catalog',
    'x': 'tables::mag_g',
    'y': 'tables::mag_g-mag_r',  # Color (G-R)
    'mode': 'markers',
    'type': 'scatter',
    'marker': {'size': 4, 'color': 'blue'}
}

layout1 = {
    'title': 'Galaxy Color-Magnitude Diagram',
    'xaxis': {'title': 'G-band magnitude'},
    'yaxis': {'title': 'G-R color'}
}

fc.show_chart(layout=layout1, data=[trace1])
print("Color-magnitude diagram created")
```

```python
# Create a histogram of redshifts
trace2 = {
    'type': 'fireflyHistogram',
    'name': 'redshift_distribution',
    'marker': {'color': 'rgba(255, 0, 0, 0.7)'},
    'firefly': {
        'tbl_id': 'galaxy_catalog',
        'options': {
            'algorithm': 'fixedSizeBins',
            'fixedBinSizeSelection': 'numBins',
            'numBins': 20,
            'columnOrExpr': 'redshift'
        }
    }
}

layout2 = {
    'title': 'Redshift Distribution',
    'xaxis': {'title': 'Redshift'},
    'yaxis': {'title': 'Number of galaxies'}
}

fc.show_chart(layout=layout2, data=[trace2])
print("Redshift histogram created")
```

## Cleanup


Clean up the temporary parquet file:


```python
# Clean up the local parquet file
if os.path.exists(parquet_file):
    os.remove(parquet_file)
    print(f"Cleaned up local file: {parquet_file}")

print("\n=== SUMMARY ===")
print("✓ Created sample parquet file with astronomical data")
print("✓ Uploaded parquet file to Firefly server")
print("✓ Displayed table with metadata in Firefly viewer")
print("✓ Created color-magnitude diagram")
print("✓ Created redshift histogram")
print("✓ Cleaned up temporary files")
print("\nThe parquet file data is now available in the Firefly viewer for interactive exploration!")
```
